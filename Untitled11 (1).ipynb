```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read Eagle image as grayscale
img = cv2.imread('eagle (2).png', 0)

# 2. Print width, height and channel
h, w = img.shape
print("Width:", w, "Height:", h, "Channels: 1")

# 3. Display grayscale image
plt.imshow(img, cmap='gray')
plt.show()

# 4. Save image as PNG
cv2.imwrite('eagle (2).png', img)

# 5. Read saved image as color
color = cv2.imread('eagle (2).png')
color = cv2.cvtColor(color, cv2.COLOR_BGR2RGB)

# 6. Display color image and print dimensions
print("Color image shape:", color.shape)
plt.imshow(color)
plt.show()

# 7. Crop eagle (adjust coordinates if needed)
crop = color[100:400, 100:400]

# 8. Resize by 2x
resize = cv2.resize(crop, None, fx=2, fy=2)

# 9. Flip horizontally
flip = cv2.flip(resize, 1)
plt.imshow(flip)
plt.show()

# 10. Read Apollo image
apollo = cv2.imread('applo-11-launch.png')

# 11. Add centered text at the bottom
h, w = apollo.shape[:2]
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN
size = cv2.getTextSize(text, font_face, 1, 1)[0]
x = (w - size[0]) // 2
cv2.putText(apollo, text, (x, h-20), font_face, 1,
            (255, 255, 255), 2)

# 12. Draw magenta rectangle
cv2.rectangle(apollo, (100, 50), (400, h-50), (255, 0, 255), 2)

# 13. Display annotated image
plt.imshow(cv2.cvtColor(apollo, cv2.COLOR_BGR2RGB))
plt.show()

# 14. Read Boy image
img = cv2.imread('boy.png')

# 15. Create matrix of ones
matrix_ones = np.ones(img.shape, dtype=np.float64) * 50
matrix = matrix_ones.astype(np.uint8)

# 16. Create brighter and darker images
img_brighter = cv2.add(img, matrix)
img_darker = cv2.subtract(img, matrix)

# 17. Display brightness images
for x in [img, img_darker, img_brighter]:
    plt.imshow(cv2.cvtColor(x, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()

# 18. Modify contrast
matrix1 = np.ones(img.shape, dtype=np.float64) * 1.1
matrix2 = np.ones(img.shape, dtype=np.float64) * 1.2

img_higher1 = cv2.multiply(img.astype(np.float64), matrix1)
img_higher2 = cv2.multiply(img.astype(np.float64), matrix2)

img_higher1 = np.clip(img_higher1, 0, 255).astype(np.uint8)
img_higher2 = np.clip(img_higher2, 0, 255).astype(np.uint8)
img_lower = cv2.convertScaleAbs(img, alpha=0.8)

# 19. Display contrast images
for x in [img, img_lower, img_higher1, img_higher2]:
    plt.imshow(cv2.cvtColor(x, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()

# 20. Split B, G, R channels
B, G, R = cv2.split(img)
for x in [B, G, R]:
    plt.imshow(x, cmap='gray')
    plt.axis('off')
    plt.show()

# 21. Merge and display RGB image
merged_rgb = cv2.merge([R, G, B])
plt.imshow(merged_rgb)
plt.title("Merged RGB Image")
plt.axis('off')
plt.show()

# 22. Split H, S, V channels
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(hsv)
for x in [H, S, V]:
    plt.imshow(x, cmap='gray')
    plt.axis('off')
    plt.show()

# 23. Merge and display HSV image
merged_hsv = cv2.merge([H, S, V])
merged_rgb = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)
plt.imshow(merged_rgb)
plt.title("Merged HSV Image")
plt.axis('off')
plt.show()
```
